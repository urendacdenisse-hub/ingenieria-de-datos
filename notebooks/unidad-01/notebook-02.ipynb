# Archivos `.txt`

Existen múltiples formas de leer archivos `.txt` en Python. En esta sección utilizaremos `open()` como una de ellas.

```python
# ==========================================
# 1. Abrir y leer el archivo
# ==========================================

ruta = "ingenieria-de-datos/datos/entrada/poema.txt"

with open(ruta, mode="r", encoding="utf-8") as archivo:
    texto = archivo.read()

# ==========================================
# 2. Mostrar el contenido
# ==========================================

print(texto)
```

Además de leer archivos, Python también permite crear nuevos archivos de texto y guardar información en ellos.

```python
# ==========================================
# 1. Crear el contenido del archivo
# ==========================================

nombre = "Escribe aquí tu nombre"

mensaje = f"""
Bienvenido al curso de Ingeniería de Datos.

Los archivos de texto son una de las formas más sencillas de almacenar
información. A diferencia de otros formatos, contienen únicamente texto
plano, por lo que son fáciles de crear, leer y compartir entre diferentes
programas y sistemas operativos.

En este ejemplo, Python creará un archivo de texto con este mensaje.

Este archivo fue creado desde Python por {nombre}.
"""

print(mensaje)
```

```python
# ==========================================
# 2. Crear y escribir el archivo
# ==========================================

ruta = "ingenieria-de-datos/datos/salida/mensaje.txt"

with open(ruta, mode="w", encoding="utf-8") as archivo:
    archivo.write(mensaje)

# ==========================================
# 3. Confirmar la operación
# ==========================================

print("Archivo creado correctamente.")
```

# Archivos CSV

En esta sección utilizaremos el archivo `clientes.csv` para aprender cómo leer, explorar y manipular datos tabulares.

```python
import pandas as pd

# ==========================================
# Lectura de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/entrada/clientes.csv"

clientes = pd.read_csv(ruta)
```

Después de cargar un archivo CSV, es buena práctica inspeccionar su contenido para confirmar que los datos se importaron correctamente y conocer su estructura general.

```python
# ==========================================
# Inspección del archivo
# ==========================================

clientes.info()   # Resume la estructura del DataFrame
```

**Observación**

Un archivo CSV puede almacenar diferentes tipos de información dentro de un mismo conjunto de datos. En este ejemplo encontramos
- números enteros (`int64`), 
- números decimales (`float64`),
- texto (`str`) y 
- valores lógicos (`bool`).

Además, observa que elementos como imágenes, audios o videos no se almacenan directamente en el archivo, sino como texto que representa su nombre, ruta o ubicación.

```python
# ==========================================
# Visualizacion de primeros registros
# ==========================================

clientes.head(n=5)
```

Al igual que un archivo de texto, la información almacenada en un **DataFrame** puede modificarse y posteriormente guardarse en un nuevo archivo CSV. En este ejemplo únicamente crearemos una copia del archivo para ilustrar cómo almacenar datos en este formato.

```python
# ==========================================
# Crear una copia del DataFrame
# ==========================================

clientes_copia = clientes.copy()

# ==========================================
# Modificar los datos
# ==========================================

clientes_copia["edad"] = clientes_copia["edad"] + 5
clientes_copia.head(n=5)
```

```python
# ==========================================
# Almacenamiento de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/salida/clientes_modificado.csv"

clientes_copia.to_csv(ruta, index=False)

print("Archivo creado correctamente.")
```

# Archivos Excel

Los archivos **Excel** (`.xlsx`) permiten almacenar información en un formato tabular similar a un archivo CSV, con la diferencia de que pueden contener **múltiples hojas de cálculo**, formatos, fórmulas y otros elementos propios de Microsoft Excel. En Python, un DataFrame puede almacenarse fácilmente como un archivo de Excel.

```python
# ==========================================
# Almacenamiento de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/salida/clientes.xlsx"

with pd.ExcelWriter(ruta) as archivo:
    clientes.to_excel(archivo, sheet_name="Original", index=False)
    clientes_copia.to_excel(archivo, sheet_name="Modificado", index=False)

print("Archivo creado correctamente.")
```

```python
# ==========================================
# Lectura de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/entrada/empresa.xlsx"

clientes = pd.read_excel(ruta, sheet_name="Clientes")
proveedores = pd.read_excel(ruta, sheet_name="Proveedores")
```

# Archivos JSON

```python
import json

# ==========================================
# Lectura de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/entrada/chats.json"

with open(ruta, mode="r", encoding="utf-8") as archivo:
    chats = json.load(archivo)  # Carga el contenido del archivo JSON

chats
```

Una vez cargado el archivo, es posible acceder a la información utilizando índices de las listas y claves del diccionario.

```python
# ==========================================
# Acceso al primer chat
# ==========================================

chats[0]
```

```python
# ==========================================
# Acceso a partes especificas del primer chat
# ==========================================

print(chats[0]["usuario"])   # Muestra el nombre del usuario
print(chats[0]["mensaje"])   # Muestra el mensaje
print(chats[0]["fecha"])     # Muestra la fecha
print(chats[0]["hora"])      # Muestra la hora
```

```python
# ==========================================
# Creacion de un nuevo chat
# ==========================================

nuevo_mensaje = {
    "fecha": "2026-07-10",
    "hora": "10:49:05",
    "usuario": "Denisse",
    "mensaje": "Hola a todos."
}

chats.append(nuevo_mensaje)   # Agrega un nuevo mensaje

chats
```

Hasta ahora trabajamos con un archivo JSON existente. Sin embargo, también es posible crear un archivo JSON desde cero utilizando estructuras de datos de Python. En el siguiente ejemplo construiremos un pequeño catálogo de libros.

```python
# ==========================================
# Creacion de los datos
# ==========================================

libros = [
    {
        "titulo": "Introducción a la Ingeniería de Datos",
        "autor": "Denisse Urenda",
        "anio": 2026,
        "disponible": True
    },
    {
        "titulo": "Python para Ciencia de Datos",
        "autor": "Juan Pérez",
        "anio": 2024,
        "disponible": False
    }
]
```

Una vez creada la estructura de datos, basta con utilizar la función `json.dump()` para almacenarla en un archivo JSON.

```python
# ==========================================
# Almacenamiento de datos
# ==========================================

ruta = "ingenieria-de-datos/datos/salida/libros.json"

with open(ruta, mode="w", encoding="utf-8") as archivo:
    json.dump(libros, archivo)

print("Archivo creado correctamente.")
```

# Archivos de imagen

Las imágenes también son una forma de almacenar datos. En Python podemos leerlas, realizar diferentes transformaciones y guardar el resultado en un nuevo archivo.

En este ejemplo utilizaremos la librería **Pillow (`PIL`)** para abrir una imagen, convertirla a escala de grises y guardar la imagen modificada.

```python
from PIL import Image

# Leer la imagen
ruta = "ingenieria-de-datos/datos/entrada/paisaje.jpg"

imagen = Image.open(ruta)
imagen
```

```python
# Características de la imagen (atributos)
print(imagen.format, imagen.size, imagen.mode)
```

```python
# Modificar la imagen: convertir a escala de grises

imagen_gris = imagen.convert("L")
imagen_gris
```

```python
# Modificar la imagen: cambiar tamaño de la imagen

imagen_redimensionada = imagen.resize((150,100))
imagen_redimensionada
```

```python
# Modificar la imagen: 
# convertir a escala de grises y cambiar tamaño de la imagen

nueva_imagen = imagen.convert("L").resize((150,100))
nueva_imagen
```

**Nota**: ¿Quieres aprender más sobre cómo manipular imágenes en Python?  Consulta el [tutorial oficial de Pillow](https://pillow.readthedocs.io/en/stable/handbook/tutorial.html).

```python
# Guardar la imagen modificada

ruta = "ingenieria-de-datos/datos/salida/imagen_transformada.jpg"
nueva_imagen.save(ruta)
```

# Archivos HTML

Las páginas web utilizan HTML para organizar y representar su contenido. En Python podemos solicitar una página web, obtener su código HTML y almacenarlo para utilizarlo posteriormente.

En este ejemplo utilizaremos la librería `requests` para obtener el contenido HTML de una página web y guardarlo en un archivo.

```python
import requests

url = "https://developer.mozilla.org/es/docs/Web/HTML"

respuesta = requests.get(url)

html = respuesta.text

print(type(html))
```

```python
# Modificar el html

html_modificado = html.replace(
    "<title>HTML: Lenguaje de Marcado de Hipertexto | MDN</title>",
    "<title>Mi página modificada</title>"
)
```

```python
# Guardar el html modificado

ruta = "ingenieria-de-datos/datos/salida/pagina_modificada.html"

with open(ruta, "w", encoding="utf-8") as archivo:
    archivo.write(html_modificado)
```

# Archivos PDF

Los archivos PDF pueden contener texto, imágenes y otros elementos en un formato diseñado para conservar su presentación. En Python podemos leer un PDF y extraer información de su contenido utilizando librerías especializadas.

En este ejemplo utilizaremos **PyPDF2** para leer un archivo PDF y extraer su texto.

```python
!pip install PyPDF2
```

```python
from PyPDF2 import PdfReader

ruta = "ingenieria-de-datos/datos/entrada/syllabus.pdf"

lector = PdfReader(ruta)

texto = lector.pages[0].extract_text()

print(texto)
```

```python
# Modificar el texto del pdf

texto_modificado = texto.replace("Ingeniería de datos", "Data Engineering")
texto_modificado
```

```python
!pip install reportlab
```

```python
# Guardar un pdf

from reportlab.platypus import SimpleDocTemplate, Paragraph

ruta = "ingenieria-de-datos/datos/salida/syllabus_modificado.pdf"

pdf = SimpleDocTemplate(ruta)
pdf.build([Paragraph(texto_modificado)])
```

```python
# Guardar un pdf con texto formateado

texto_formateado = texto_modificado.replace("\n", "<br/>")

pdf = SimpleDocTemplate(ruta)
pdf.build([Paragraph(texto_formateado)])
```

**Nota:** Crear un archivo PDF es sencillo, pero conservar o construir una presentación adecuada puede requerir trabajo adicional. Un PDF puede contener información sobre texto, posiciones, fuentes, imágenes y otros elementos de diseño.

Al extraer únicamente el texto de un PDF, parte de esta información se pierde. Por esta razón, el documento generado puede verse diferente al original.
