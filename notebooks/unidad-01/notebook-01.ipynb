# ¿Qué son los datos?

Un **dato** es una representación de alguna característica, evento o fenómeno del mundo real.

Los datos permiten describir personas, objetos, lugares, mediciones, imágenes, sonidos, videos y prácticamente cualquier elemento que pueda observarse o registrarse.

En Ingeniería de Datos, los datos representan la materia prima con la que posteriormente se realizan procesos de almacenamiento, transformación y análisis para obtener información útil.

# Tipos de datos

## Dato numérico

Los datos numéricos representan cantidades que pueden **medirse** o **contarse**. Algunos ejemplos son la edad de una persona, la temperatura de una ciudad o el precio de un producto.


```python
edad = 22
temperatura = 36.8
precio = 1499.99

print(edad, type(edad))
print(temperatura, type(temperatura))
print(precio, type(precio))
```

**Observación**: Aunque ambos representan valores numéricos, Python distingue entre dos tipos principales:

- `int` (*integer*): representa números enteros.
- `float` (*floating point*): representa números con parte decimal.

Una de las principales características de los datos numéricos es que permiten realizar operaciones matemáticas como sumas, restas, multiplicaciones y divisiones.

Por ejemplo, podemos calcular el precio total de una compra o la temperatura promedio de una ciudad.

```python
camisa = 350.00
pantalon = 620.00
tenis = 1299.99

subtotal = camisa + pantalon + tenis
iva = subtotal * 0.16
total = subtotal + iva

print(f"Subtotal: ${subtotal}")
print(f"Iva: ${iva}")
print(f"Total: ${total}")
```

```python
from recursos.codigo.tickets import generar_ticket

productos = [
    ("Camisa", 3, 350.00),
    ("Pantalón", 3, 620.00),
    ("Tenis deportivos", 1, 1299.99),
    ("Calcetines", 3, 89.90),
    ("Celular", 10, 5700.00)
]

ticket = generar_ticket(productos)
print(ticket)
```

## Dato de tipo texto

Los datos de texto permiten representar palabras, frases o cualquier secuencia de caracteres. Son ampliamente utilizados para almacenar nombres, direcciones, correos electrónicos, comentarios y otra información descriptiva.

Observemos algunos ejemplos utilizando Python.

```python
nombre = "Ana Martínez"
ciudad = "Ciudad Juárez"
correo = "ana@email.com"

print(nombre, type(nombre))
print(ciudad, type(ciudad))
print(correo, type(correo))
```

**Observación**

Python identifica los datos de tipo texto como `str` (abreviatura de *string*). Aunque suelen representar texto, también pueden utilizarse para almacenar cualquier secuencia de caracteres, como números telefónicos, códigos postales o identificadores.

```python
from recursos.codigo.promocion import generar_correo_promocional

mensaje = generar_correo_promocional(
    nombre = "Marcos Jiménez",
    correo = "marcos.jimenez@email.com",
    ciudad = "Ciudad Juárez"
)
print(mensaje)
```

## Datos tipo fecha y hora

Los datos de tipo fecha y hora permiten representar momentos específicos en el tiempo, como una fecha, una hora o ambas combinadas. Son fundamentales para registrar cuándo ocurrió un evento.

| Valor                 | Descripción                     |
| --------------------- | ------------------------------- |
| `2026-07-09`          | Fecha                           |
| `14:35:20`            | Hora                            |
| `2026-07-09 14:35:20` | Fecha y hora                    |



```python
from datetime import datetime

ahora = datetime.now()

print(ahora)
print(ahora.date())
print(ahora.time())
```

**Observación**

En Python, el tipo `datetime` almacena tanto la fecha como la hora. Cuando es necesario, pueden obtenerse por separado utilizando los métodos `.date()` y `.time()`.

```python
from recursos.codigo.chat import buscar_chat

historial = buscar_chat(fecha = "2026-07-08")
print(historial)
```

```python
historial = buscar_chat(fecha = "2026-07-09")
print(historial)
```

## Datos multimedia

Los datos multimedia permiten almacenar información visual o auditiva. A diferencia de los datos numéricos o de texto, suelen ocupar un mayor espacio de almacenamiento y requieren herramientas especializadas para su procesamiento y análisis.

| Tipo   | Ejemplo                                            |
| ------ | -------------------------------------------------- |
| Imagen | Fotografía, radiografía, logotipo                  |
| Audio  | Canción, nota de voz, llamada telefónica           |
| Video  | Grabación de seguridad, película, videoconferencia |


**Observación**

Aunque imágenes, audios y videos pueden visualizarse como contenido multimedia, internamente también son datos digitales y pueden ser procesados mediante programas para aplicar filtros, modificar su tamaño, extraer información o realizar análisis.

```python
from recursos.codigo.imagenes import descargar_imagen, mostrar_imagen

# Asegurate de que la direccion url termine alguna de las siguientes extensiones
# ".jpg", ".jpeg", ".png", ".gif", ".bmp", ".webp"

ruta_original = descargar_imagen(
    url = "https://upload.wikimedia.org/wikipedia/commons/e/ed/Cynomys_ludovicianus.005_-_Faunia.JPG",
    nombre_archivo = "perro_pradera.png"
)

print(ruta_original)
```

```python
mostrar_imagen(origen = ruta_original)
```

```python
url = "https://upload.wikimedia.org/wikipedia/commons/3/3f/Fronalpstock_big.jpg"

mostrar_imagen(origen = url)
```

Las imágenes también pueden procesarse mediante software. Es posible aplicar transformaciones como *filtros*, *rotaciones* o *cambios de tamaño* para adaptarlas a diferentes necesidades.

```python
from recursos.codigo.imagenes import aplicar_filtro

ruta_modificada = aplicar_filtro(
    ruta_imagen = ruta_original,
    filtro = "contorno",
    nombre_salida = "perro_contorno.png"
)

mostrar_imagen(origen = ruta_modificada)
```

```python
ruta_modificada = aplicar_filtro(
    ruta_imagen = ruta_original,
    filtro = "bordes",
    nombre_salida = "perro_bordes.png"
)

mostrar_imagen(origen = ruta_modificada)
```

```python
from recursos.codigo.imagenes import redimensionar_imagen

ruta_modificada = redimensionar_imagen(
    ruta_imagen = ruta_original,
    ancho = 350,
    alto = 500,
    nombre_salida = "perro_escalado.png"
)

mostrar_imagen(origen = ruta_modificada)
```

```python
from recursos.codigo.imagenes import rotar_imagen

ruta_modificada = rotar_imagen(
    ruta_imagen = ruta_original,
    grados = 90,
    nombre_salida = "perro_rotado.png"
)

mostrar_imagen(origen = ruta_modificada)
```
