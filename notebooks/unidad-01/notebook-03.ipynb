# Archivos

Los archivos son una de las formas más simples de almacenar e intercambiar datos. Como vimos anteriormente, pueden utilizar diferentes formatos dependiendo del tipo de información que contienen.

En este ejemplo leeremos directamente un archivo CSV.

```python
import pandas as pd

clientes = pd.read_csv("ingenieria-de-datos/datos/entrada/clientes.csv")

clientes.head()
```

# Base de datos

Una base de datos permite almacenar y organizar información para que pueda ser consultada de manera eficiente.

En este ejemplo utilizaremos **SQLite** para conectarnos a una base de datos existente y consultar los datos almacenados en ella.

```python
import sqlite3

ruta = "ingenieria-de-datos/datos/entrada/empresa.db"

conexion = sqlite3.connect(ruta)
```

```python
consulta = """
SELECT name
FROM sqlite_master
WHERE type='table';
"""

pd.read_sql(consulta, conexion)
```

```python
consulta = "SELECT * FROM clientes"

clientes = pd.read_sql(consulta, conexion)

clientes
```

```python
conexion.close()
```

**Nota**: A diferencia de otros formatos de almacenamiento, las bases de datos se consultan mediante **SQL** (Structured Query Language), un lenguaje diseñado específicamente para crear, consultar y modificar información. En la siguiente unidad aprenderemos los fundamentos de este lenguaje.

# APIs

Una API permite que una aplicación solicite información o servicios a otra aplicación.

A diferencia de un archivo local, los datos se solicitan a través de una dirección y el servidor devuelve una **respuesta**.

## JSONPlaceholder

**[JSONPlaceholder](https://jsonplaceholder.typicode.com/)** es un servicio web gratuito que simula una API real utilizando datos ficticios.

Contiene recursos como usuarios, publicaciones, comentarios y tareas, y permite realizar solicitudes para obtener estos datos en formato JSON.

En este ejemplo consultaremos la información de los usuarios disponibles.

```python
import requests

url = "https://jsonplaceholder.typicode.com/users"

# Hacemos la solicitud
respuesta = requests.get(url)

# Convertimos la respuesta a JSON
datos = respuesta.json()

datos
```

```python
# Convertimos el JSON en DataFrame de pandas
usuarios = pd.DataFrame(datos)

usuarios.head()
```

## Open-Meteo

**[Open-Meteo](https://open-meteo.com/)** es un servicio que proporciona información meteorológica y pronósticos del tiempo mediante una API.

Esta API permite indicar mediante **parámetros** la ubicación y la información meteorológica que queremos consultar.

En este ejemplo solicitaremos la temperatura actual en Ciudad Juárez.

```python
# Dirección de la API
url = "https://api.open-meteo.com/v1/forecast"

# Definimos los parámetros de la consulta
parametros = {
    "latitude": 31.69,
    "longitude": -106.42,
    "current": "temperature_2m"
}

# Hacemos la solicitud enviando los parámetros
respuesta = requests.get(url, params=parametros)

# Convertimos la respuesta a JSON
datos = respuesta.json()

datos
```

Los parámetros también pueden escribirse directamente en la URL de la solicitud.

En este caso, `?` indica el inicio de los parámetros y `&` permite separar cada uno.

```python
# Construimos manualmente la URL con los parámetros
url = "https://api.open-meteo.com/v1/forecast?latitude=31.69&longitude=-106.42&current=temperature_2m"

# Hacemos la solicitud
respuesta = requests.get(url)

# Convertimos la respuesta a JSON
datos = respuesta.json()

datos
```

## NASA Open APIs

**[NASA Open APIs](https://api.nasa.gov/)** proporciona acceso a diferentes datos de la NASA mediante APIs públicas.

A diferencia de los ejemplos anteriores, algunas APIs requieren una **clave de acceso (API key)** para identificar al usuario que realiza la solicitud.

NASA proporciona una clave de demostración llamada `DEMO_KEY` que podemos utilizar para realizar consultas sencillas.

En este ejemplo consultaremos la imagen astronómica del día.

```python
# Dirección de la API
url = "https://api.nasa.gov/planetary/apod"

# Definimos los parámetros de la consulta
parametros = {
    "api_key": "DEMO_KEY"
}

# Hacemos la solicitud
respuesta = requests.get(url, params=parametros)

# Convertimos la respuesta a JSON
datos = respuesta.json()

datos
```

**Nota**: Puedes consultar la lista de parametros disponibles [aquí](https://api.nasa.gov/?search=planetary).

```python
# Extraemos algunos datos de la respuesta
print("Título:", datos["title"])
print("Fecha:", datos["date"])
print("Tipo:", datos["media_type"])
```

```python
from IPython.display import Image, display

display(Image(url=datos["url"]))
```

# Páginas web

Una página web también puede utilizarse como fuente de datos. Al realizar una solicitud, normalmente recibimos un documento **HTML** que contiene tanto el contenido como la estructura de la página.

A diferencia de una API, la información no necesariamente se encuentra preparada para ser utilizada directamente como datos. Por ello, debemos identificar y extraer los elementos que nos interesan.

A este proceso se le conoce como **web scraping**.

### Books to Scrape

**Books to Scrape** es un sitio creado específicamente para practicar técnicas de web scraping.

La página contiene un catálogo ficticio de libros con información como título, precio, disponibilidad y calificación.

En este ejemplo extraeremos información directamente del contenido HTML de la página.

```python
import requests

url = "https://books.toscrape.com/"

# Hacemos la solicitud
respuesta = requests.get(url)

# Obtenemos el contenido HTML
html = respuesta.text

html[:1000]
```

**BeautifulSoup** es una biblioteca de Python que permite **interpretar y recorrer documentos HTML** para localizar y extraer elementos específicos de una página web.

Facilita tareas como buscar etiquetas, acceder a sus atributos y extraer su contenido.

```python
!pip install bs4
```

```python
from bs4 import BeautifulSoup

# Interpretamos la estructura del documento HTML
pagina = BeautifulSoup(html, "html.parser")
```

```python
# Buscamos el primer título h3
titulo = pagina.find("h3")

titulo
```

```python
# Extraemos únicamente el texto
titulo.get_text()
```

```python
# Buscamos todos los libros de la página
libros = pagina.find_all("article", class_="product_pod")

libros
```

```python
# Extraccion de información

titulo = libros[0].find("h3").get_text()
precio = libros[0].find("p", class_="price_color").get_text()
disponibilidad = libros[0].find("p", class_="instock availability").get_text()

print(titulo)
print(precio)
print(disponibilidad)
```

```python
# Formateo de informacion
precio = float(precio[2:])
disponibilidad = disponibilidad.strip()

print(precio)
print(disponibilidad)
```

```python
lista_articulos = []

for libro in libros:
    titulo = libro.find("h3").get_text()
    precio = float(libro.find("p", class_="price_color").get_text()[2:])
    disponibilidad = libro.find("p", class_="instock availability").get_text().strip()

    lista_articulos.append((titulo, precio, disponibilidad))
    
```

```python
lista_articulos
```

```python
articulos = pd.DataFrame(
    lista_articulos, 
    columns=["titulo", "precio", "disponibilidad"]
)

articulos
```
